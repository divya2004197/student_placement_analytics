```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
sns.countplot(
    x='Placement_Status',
    data=df
)
plt.title('placement status distribution')
plt.show()
```

```python
sns.histplot(
    df['CGPA'],
    bins=20
)
plt.title('CGPA Distribution')
plt.show()
```

```python
sns.countplot(
    y='Branch',
    data=df
)
```

```python
df.groupby('Placement_Status')[[
    'CGPA',
    'Coding_Skills',
    'Communication_Skills',
    'Aptitude_Test_Score'
]].mean()
```

```python
branch_placement=pd.crosstab(
    df['Branch'],
    df['Placement_Status']
)
branch_placement
```

```python
branch_placement.plot(kind='bar',figsize=(10,5))
plt.title("Branch-wise placements")
plt.show()
```

```python
pd.crosstab(
    df['Gender'],
    df['Placement_Status']
)
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('correlation Heatmap')
plt.show()
```
